```python
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
from IPython.display import display, clear_output
import time
from test_env import SimpleEnv

env = SimpleEnv(render_mode='rgb_array')
obs, _ = env.reset(seed=42)

def show_frame(title=''):
    frame = env.render()
    fig, ax = plt.subplots(figsize=(5, 5))
    ax.imshow(frame)
    ax.set_title(title or f'pos={env.agent_pos} dir={env.agent_dir}')
    ax.axis('off')
    plt.tight_layout()
    plt.show()

show_frame('Initial state')
```

```python
# Actions: 0=turn left, 1=turn right, 2=move forward, 3=pick up, 4=drop, 5=toggle (open door)
ACTION_NAMES = ['turn_left', 'turn_right', 'forward', 'pickup', 'drop', 'toggle']

def step(action: int):
    obs, reward, terminated, truncated, info = env.step(action)
    show_frame(f'action={ACTION_NAMES[action]}  reward={reward:.2f}  done={terminated}')
    return obs, reward, terminated, truncated

# Example: move forward
step(2)
```

```python
# Step through a sequence of actions and view each frame
obs, _ = env.reset(seed=42)

actions = [2, 2, 1, 2, 2, 2, 3, 1, 1, 2, 2, 2, 5]  # walk, pick up key, open door

fig, axes = plt.subplots(2, 7, figsize=(20, 6))
axes = axes.flatten()

frame = env.render()
axes[0].imshow(frame)
axes[0].set_title('start')
axes[0].axis('off')

for i, action in enumerate(actions, 1):
    env.step(action)
    frame = env.render()
    axes[i].imshow(frame)
    axes[i].set_title(ACTION_NAMES[action])
    axes[i].axis('off')

plt.tight_layout()
plt.show()
```

```python
# Save a GIF of a random agent episode
import imageio

obs, _ = env.reset(seed=42)
frames = [env.render()]

for _ in range(60):
    action = env.action_space.sample()
    obs, reward, terminated, truncated, _ = env.step(action)
    frames.append(env.render())
    if terminated or truncated:
        break

imageio.mimsave('episode.gif', frames, fps=4)
print(f'Saved {len(frames)} frames → episode.gif')
```
